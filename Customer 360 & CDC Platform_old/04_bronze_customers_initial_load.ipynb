```python
source_path = (
    "/Volumes/customer360_dev/staging/"
    "source_files/customers/initial/"
)
```

### Read the source files

```python
from pyspark.sql import functions as F

customers_source_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", False)
    .csv(source_path)
)
```

### Add Bronze metadata

```python
customers_bronze_df = (
    customers_source_df
    .withColumn(
        "_ingestion_timestamp",
        F.current_timestamp()
    )
    .withColumn(
        "_source_file",
        F.col("_metadata.file_path")
    )
    .withColumn(
        "_source_system",
        F.lit("CRM")
    )
    .withColumn(
        "_load_type",
        F.lit("INITIAL")
    )
)
```

### Write to Delta

```python
(
    customers_bronze_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "customer360_dev.bronze.customers"
    )
)
```

### Verify row count

```python
bronze_count = spark.table(
    "customer360_dev.bronze.customers"
).count()

print(f"Bronze customer rows: {bronze_count:,}")
```

```python
display(
    spark.table(
        "customer360_dev.bronze.customers"
    )
    .select(
        "customer_id",
        "email",
        "_source_file",
        "_source_system",
        "_load_type",
        "_ingestion_timestamp"
    )
    .limit(20)
)
```
