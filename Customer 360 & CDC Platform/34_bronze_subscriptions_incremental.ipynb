```python
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType
import uuid

# ============================================================
# 1. Source / Target
# ============================================================

source_path = (
    "/Volumes/customer360_dev/staging/"
    "source_files/subscriptions/incremental/"
)

target_table = "customer360_dev.bronze.subscriptions"

run_id = str(uuid.uuid4())

print("Subscription Pipeline Run ID:", run_id)
```

```python

# ============================================================
# 2. Explicit Subscription Schema
# ============================================================

subscription_schema = StructType([
    StructField("subscription_id", StringType(), True),
    StructField("customer_id", StringType(), True),
    StructField("plan_code", StringType(), True),
    StructField("subscription_status", StringType(), True),
    StructField("start_date", StringType(), True),
    StructField("end_date", StringType(), True),
    StructField("billing_cycle", StringType(), True),
    StructField("monthly_fee", StringType(), True),
    StructField("auto_renew", StringType(), True),
    StructField("payment_method_type", StringType(), True),
    StructField("trial_flag", StringType(), True),
    StructField("cancel_reason", StringType(), True),
    StructField("created_at", StringType(), True),
    StructField("updated_at", StringType(), True)
])
```

```python

# ============================================================
# 3. Read Incremental CSV Files
# ============================================================

incremental_df = (
    spark.read
    .option("header", True)
    .schema(subscription_schema)
    .csv(source_path)
)
```

```python

# ============================================================
# 4. Add Bronze Metadata
# ============================================================

bronze_incremental_df = (
    incremental_df
    .withColumn(
        "_ingestion_timestamp",
        F.current_timestamp()
    )
    .withColumn(
        "_source_file",
        F.col("_metadata.file_path")
    )
    .withColumn(
        "_source_system",
        F.lit("SUBSCRIPTION_PLATFORM")
    )
    .withColumn(
        "_load_type",
        F.lit("INCREMENTAL")
    )
    .withColumn(
        "_pipeline_run_id",
        F.lit(run_id)
    )
)

```

```python

# ============================================================
# 5. Preview
# ============================================================

display(
    bronze_incremental_df.limit(20)
)
```

```python

# ============================================================
# 6. Count Incoming Rows
# ============================================================

incoming_count = bronze_incremental_df.count()

print(
    f"Incoming subscription incremental rows: "
    f"{incoming_count:,}"
)
```

```python

# ============================================================
# 7. Append to Bronze
# ============================================================

if incoming_count > 0:

    (
        bronze_incremental_df
        .write
        .format("delta")
        .mode("append")
        .option("mergeSchema", "true")
        .saveAsTable(
            target_table
        )
    )

    print(
        f"Subscription incremental batch loaded. "
        f"run_id = {run_id}"
    )

else:

    print(
        "No incremental subscription records found. "
        "Bronze write skipped."
    )
```

```python
if incoming_count > 0:

    written_count = (
        spark.table(target_table)
        .filter(
            F.col("_pipeline_run_id") == run_id
        )
        .count()
    )

else:

    written_count = 0

print(f"Incoming rows : {incoming_count:,}")
print(f"Written rows  : {written_count:,}")
```
