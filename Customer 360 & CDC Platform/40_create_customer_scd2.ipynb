```python
%sql
CREATE TABLE IF NOT EXISTS customer360_dev.gold.dim_customer
(
    customer_sk BIGINT GENERATED ALWAYS AS IDENTITY,

    customer_id STRING,

    first_name STRING,
    last_name STRING,
    email STRING,
    phone_number STRING,
    date_of_birth DATE,
    gender STRING,
    signup_date DATE,
    customer_status STRING,
    loyalty_tier STRING,
    preferred_language STRING,

    effective_from TIMESTAMP,
    effective_to TIMESTAMP,
    is_current BOOLEAN,

    source_updated_at TIMESTAMP,

    CONSTRAINT pk_dim_customer
        PRIMARY KEY (customer_sk)
        NOT ENFORCED
)
USING DELTA;
```

```python
%sql
DESCRIBE TABLE customer360_dev.gold.dim_customer;
```
